# AAL Q4 2020 Sales Analysis

## Australian Apparal Limited - Fourth Quarter Sales Performance

Let's start by understanding what we're working with. The data contains sales transactions across Australian states for the fourth quarter of 2020, broken down by time of day and customer demographics.

```python
import pandas as pd

df = pd.read_csv('../AusApparalSales4thQrt2020.csv')

for col in ['Time', 'State', 'Group']:
    df[col] = df[col].str.strip()

print(f"Dataset shape: {df.shape}")
print(f"\nFirst few records:")
print(df.head())
```

```python
print("Data Types:")
print(df.dtypes)
print(f"\nMissing Values:")
print(df.isnull().sum())
print(f"\nDataset Info:")
df.info()
```

```python
print("Quick check on states:")
print(df['State'].unique())
```

```python
sample_states = df['State'].head(20).tolist()
print("Sample state values (first 20 rows):")
for i, state in enumerate(sample_states):
    print(f"  Row {i}: '{state}'")

print("\nLooks like there's whitespace in the state names. Need to strip those.")
print("Already handled in the load step above.")
```

```python
print("Unique States:")
print(df['State'].unique())
print(f"\nUnique Groups:")
print(df['Group'].unique())
print(f"\nUnique Time Periods:")
print(df['Time'].unique())
print(f"\nTotal Records per State:")
print(df['State'].value_counts())
```

```python
print("Sales Summary Statistics:")
print(df['Sales'].describe())
print(f"\nUnits Summary Statistics:")
print(df['Unit'].describe())
```

```python
total_revenue = df['Sales'].sum()
total_units = df['Unit'].sum()
avg_transaction_value = df['Sales'].mean()

print(f"Total Q4 Revenue: ${total_revenue:,.0f}")
print(f"Total Units Sold: {total_units:,}")
print(f"Average Transaction Value: ${avg_transaction_value:,.0f}")
print(f"Number of Transactions: {len(df):,}")
```
